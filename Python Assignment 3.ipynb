### While loop & Control Statements :

```python
# Number guessing game using(while loop, else, break, continue)
import random
secret_num=random.randint(1,10)
attempts=3
while attempts>0: 
    num=int(input("guess the number"))
    if num<1 or num>10: 
        print("Your guess is out of range, please guess the number between 1 to 10")
        continue
    elif num> secret_num:
        print("Too high! try again")
    elif num< secret_num:  
        print("Too low! try again")
    else: 
        print("Congratulations!! You have guessed the correct number")
        break

        atempts-=1
        print("Attempts left:",attempts)
else: 
    print("Better luck next time!")
```

### For Loop:

```python
num=int(input("enter the number for which you want the multiplication table"))
for i in range(1,11): 
    print(f"{i} × {num} = {i*num}")
```
